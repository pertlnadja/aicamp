**Erste Analyse zur Fragestellung: "Automatisierte Zeitreihenanalyse (Trends, Drifts, Anomalien)"**

Wie kann eine KI Trends, Drifts und Anomalien in Prozessparametern entlang ganzer Prozessketten automatisch erkennen?

Beispiel: Schraub‑ und Presskurven, Prozesskennzahlen, Bauteilmerkmale.
Ziel: „Good Curve Learning“ → KI erkennt typische Kurvenverläufe und warnt live bei Abweichungen.  

Idealziel:  
•	Automatische Erkennung von Über‑/Unterdrehen, Reibungsanomalien, falschem Sitz  
•	Sofortiges Stoppen / Nacharbeiten statt „Durchschleusen“ bis EOL
Kontext & Bedeutung  
Qualitätssicherung über die gesamte Prozesskette hinweg, frühzeitige Fehlervermeidung.
