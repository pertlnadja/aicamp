**Erste Analyse zur Fragestellung: "Analyse hochdimensionaler Vibrationsdaten"**

Analyse hochdimensionaler Vibrationsdaten (2000 Variablen × ~100 Spektren pro Teil)
Konkrete Fragestellung:  
Wie können sehr große Vibrations‑/Frequenzspektren effizient modelliert und für Klassifikation/Regression genutzt werden?  
Offene Punkte:  
•	Reduktion der ca. 2000 Frequenzvariablen (z. B. PCA, Autoencoder, Spektrum als „Bild“/1D‑Signal)  
•	Ausreißererkennung in großen Datenmengen (>100.000 Datenpunkte)  
•	Identifikation irrelevanter Variablen / Feature‑Selection  
•	Evaluierung geeigneter Modelle in KNIME für 1D‑Signal‑Daten  

Kontext & Bedeutung:
Vibrationsmessungen sind ein häufiges Qualitätsmerkmal. Eine automatisierte Auswertung würde Inspektionszeiten reduzieren und Qualitätsentscheidungen objektivieren.  

Bisherige Erkenntnisse:  
Bisher nur Basisstatistiken und Visualisierungen.
